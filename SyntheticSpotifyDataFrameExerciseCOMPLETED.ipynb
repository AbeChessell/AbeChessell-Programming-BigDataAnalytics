```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter the dataframe for the three subscription types of interest
compare_df = df[df['subscription_type'].isin(['Free', 'Premium', 'Student'])]

# Set the aesthetic style for the plots
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Graph 1: Box Plot of Listening Time by Subscription Type
sns.boxplot(x='subscription_type', y='listening_time', data=compare_df, ax=axes[0, 0], palette='Set2')
axes[0, 0].set_title('Distribution of Listening Time')
axes[0, 0].set_xlabel('Subscription Type')
axes[0, 0].set_ylabel('Listening Time (mins)')

# Graph 2: Violin Plot of Songs Played Per Day
sns.violinplot(x='subscription_type', y='songs_played_per_day', data=compare_df, ax=axes[0, 1], palette='Pastel1')
axes[0, 1].set_title('Distribution of Songs Played Per Day')
axes[0, 1].set_xlabel('Subscription Type')
axes[0, 1].set_ylabel('Songs Played Per Day')

# Graph 3: Bar Plot of Churn Rate by Subscription Type
sns.barplot(x='subscription_type', y='is_churned', data=compare_df, ax=axes[1, 0], palette='muted', errorbar=None)
axes[1, 0].set_title('Churn Rate Comparison')
axes[1, 0].set_xlabel('Subscription Type')
axes[1, 0].set_ylabel('Proportion Churned')

# Graph 4: KDE Plot / Density of Age Distribution
sns.kdeplot(data=compare_df, x='age', hue='subscription_type', fill=True, common_norm=False, alpha=0.4, ax=axes[1, 1], palette='Set1')
axes[1, 1].set_title('Age Distribution Density')
axes[1, 1].set_xlabel('Age')
axes[1, 1].set_ylabel('Density')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter for Free users since Premium and Student users do not have ads
free_users_df = df[df['subscription_type'] == 'Free']

# Set up a 1x2 subplot layout
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Graph 1: Scatter plot of Ads Listened vs. Listening Time for Free Users, colored by Churn
sns.scatterplot(
    data=free_users_df,
    x='ads_listened_per_week',
    y='listening_time',
    hue='is_churned',
    palette='Set1',
    alpha=0.6,
    ax=axes[0]
)
axes[0].set_title('Free Users: Listening Time vs. Ads Listened Per Week')
axes[0].set_xlabel('Ads Listened Per Week')
axes[0].set_ylabel('Listening Time (mins)')
axes[0].legend(title='Is Churned', labels=['No', 'Yes'])

# Graph 2: Boxplot comparing Ads Listened for Churned vs. Active Free Users
sns.boxplot(
    data=free_users_df,
    x='is_churned',
    y='ads_listened_per_week',
    palette='Set2',
    hue='is_churned',
    legend=False,
    ax=axes[1]
)
axes[1].set_title('Free Users: Ads Listened Per Week by Churn Status')
axes[1].set_xlabel('Is Churned')
axes[1].set_ylabel('Ads Listened Per Week')
axes[1].set_xticklabels(['Active (0)', 'Churned (1)'])

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter for Desktop and Mobile users
device_df = df[df['device_type'].isin(['Desktop', 'Mobile','Web'])]

# Calculate exact churn rates
churn_rates = device_df.groupby('device_type')['is_churned'].mean() * 100
print("Churn Rates by Device Type:")
for device, rate in churn_rates.items():
    print(f"{device}: {rate:.2f}%")

# Plot the comparison
plt.figure(figsize=(8, 5))
sns.barplot(
    x='device_type',
    y='is_churned',
    data=device_df,
    errorbar=None,
    palette='pastel',
    hue='device_type',
    legend=False
)
plt.title('Churn Rate Comparison: Desktop vs. Mobile vs. Web users')
plt.xlabel('Device Type')
plt.ylabel('Churn Rate (Proportion)')
plt.ylim(0, 1) # Set limits to represent proportion from 0 to 1
plt.show()
```

```python
# Calculate the Pearson correlation coefficient between listening_time and is_churned
correlation = df['listening_time'].corr(df['is_churned'])
print(f"The correlation between listening time and churn status is: {correlation:.4f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Identify numerical columns to compare against churn status (excluding 'user_id' and 'is_churned')
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
numerical_cols = [col for col in numerical_cols if col not in ['user_id', 'is_churned']]

# Calculate correlation of all numerical columns with 'is_churned'
correlations = df[numerical_cols].corrwith(df['is_churned']).sort_values()

print("Correlation of numerical columns with Churn Status (is_churned):")
print(correlations)

# Plot the correlations
plt.figure(figsize=(10, 6))
sns.barplot(x=correlations.values, y=correlations.index, palette='coolwarm', hue=correlations.index, legend=False)
plt.axvline(x=0, color='black', linestyle='--', linewidth=1)
plt.title('Correlation of Numerical Features with Churn Status')
plt.xlabel('Pearson Correlation Coefficient')
plt.ylabel('Features')
plt.xlim(-0.1, 0.1)  # Setting limits to capture these small correlations clearly
plt.show()
```
